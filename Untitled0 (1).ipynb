## Q1, Q2, Q3: Read an image, check for successful loading, and display its properties.

```python
import cv2
import matplotlib.pyplot as plt

# Q1: Read an image using OpenCV
image_path = '/content/flowers.jpg'  # <---- CHANGE THIS TO YOUR IMAGE PATH
img = cv2.imread(image_path)

# Q2: Check whether an image was loaded successfully
if img is None:
    print(f"Error: Could not load image from {image_path}. Please check the path and filename.")
else:
    print(f"Image loaded successfully from {image_path}.")

    # Q3: Print the image height, width, and number of channels
    # For color images, shape will be (height, width, channels)
    # For grayscale images, shape will be (height, width)
    if len(img.shape) == 3:
        height, width, channels = img.shape
        print(f"Image dimensions: Height = {height} pixels, Width = {width} pixels, Channels = {channels}")
    elif len(img.shape) == 2:
        height, width = img.shape
        print(f"Image dimensions: Height = {height} pixels, Width = {width} pixels, Channels = 1 (grayscale)")
    else:
        print("Could not determine image dimensions.")
```

## Q4, Q5: Calculate total pixels and print data type

```python
# Q4: Calculate and print the total number of pixels in an image
# Total pixels = height * width
# For color images, it's often still considered height * width, not including channels

if img is not None:
    if len(img.shape) == 3:
        height, width, _ = img.shape
    else:
        height, width = img.shape
    total_pixels = height * width
    print(f"Total number of pixels: {total_pixels}")

    # Q5: Print the data type (`dtype`) of the image matrix
    print(f"Image data type (dtype): {img.dtype}")
else:
    print("Image not loaded, cannot calculate total pixels or dtype.")
```

## Q6: Save an image with a different filename

```python
# Q6: Read an image and save it with a different filename using OpenCV
output_image_path = 'output_flowers.jpg'

if img is not None:
    cv2.imwrite(output_image_path, img)
    print(f"Image saved successfully as {output_image_path}")
else:
    print("Image not loaded, cannot save.")
```

## Q7, Q8: Read/Convert to grayscale and display

```python
# Q7: Read an image directly in grayscale mode and display it
grayscale_img_direct = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if grayscale_img_direct is not None:
    print("Image read directly in grayscale mode.")
    plt.figure(figsize=(6, 6))
    plt.imshow(grayscale_img_direct, cmap='gray')
    plt.title('Image Read Directly in Grayscale')
    plt.axis('off') # Hide axis
    plt.show()
else:
    print("Could not read image directly in grayscale.")

# Q8: Convert a color image to grayscale using `cv2.cvtColor()`
if img is not None:
    grayscale_img_converted = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    print("Color image converted to grayscale using cv2.cvtColor().")
    plt.figure(figsize=(6, 6))
    plt.imshow(grayscale_img_converted, cmap='gray')
    plt.title('Color Image Converted to Grayscale')
    plt.axis('off') # Hide axis
    plt.show()
else:
    print("Original image not loaded, cannot convert to grayscale.")
```

## Q9: Display an image using Matplotlib and hide the axis

```python
# Q9: Display an image using Matplotlib and hide the axis
# We'll display the original color image here, ensuring it's in RGB format for Matplotlib

if img is not None:
    # OpenCV reads images as BGR, Matplotlib expects RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(8, 8))
    plt.imshow(img_rgb)
    plt.title('Original Image Displayed with Matplotlib (Axes Hidden)')
    plt.axis('off') # This line hides the axis
    plt.show()
else:
    print("Image not loaded, cannot display with Matplotlib.")
```

## Q10: Resize an image

```python
# Q10: Resize an image to 50% of its original width and height

if img is not None:
    # Get original dimensions
    original_height, original_width = img.shape[:2]

    # Calculate new dimensions (50%)
    new_width = int(original_width * 0.5)
    new_height = int(original_height * 0.5)

    # Resize the image
    resized_img = cv2.resize(img, (new_width, new_height), interpolation=cv2.INTER_AREA)

    print(f"Original dimensions: {original_width}x{original_height}")
    print(f"Resized dimensions (50%): {new_width}x{new_height}")

    # Display the resized image
    # Convert BGR to RGB for Matplotlib display
    resized_img_rgb = cv2.cvtColor(resized_img, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(8, 8))
    plt.imshow(resized_img_rgb)
    plt.title('Image Resized to 50%')
    plt.axis('off')
    plt.show()
else:
    print("Image not loaded, cannot resize.")
```

## Q11-Q18: Pixel & Image Representation (Access, Modify, Split, Merge, Min/Max/Mean)

```python
import numpy as np

# Ensure the image is loaded before proceeding
if img is None:
    print("Error: Image not loaded. Please execute cell `7ce51561` first.")
else:
    # Q11. Access and print the pixel value at a user-provided (x, y) coordinate.
    # Using example coordinates. Adjust x_coord and y_coord as needed.
    x_coord, y_coord = 50, 100 # Example coordinates

    # Check if coordinates are within image bounds
    if 0 <= y_coord < img.shape[0] and 0 <= x_coord < img.shape[1]:
        pixel_value = img[y_coord, x_coord]
        print(f"Q11: Pixel value at ({x_coord}, {y_coord}): {pixel_value}")
    else:
        print(f"Q11: Coordinates ({x_coord}, {y_coord}) are out of image bounds.")

    # Q12. Modify the intensity/value of a selected pixel and save the modified image.
    # For a color image, modifying a pixel means modifying its B, G, R components
    # For simplicity, we'll modify a small area or a single channel.
    modified_img = img.copy() # Create a copy to avoid modifying the original 'img'
    if len(modified_img.shape) == 3: # Color image
        modified_img[y_coord, x_coord] = [0, 0, 255] # Set to bright Red (BGR format)
        print(f"Q12: Modified pixel at ({x_coord}, {y_coord}) to Red. Saving 'modified_pixel.jpg'.")
    else: # Grayscale image
        modified_img[y_coord, x_coord] = 200 # Set to a bright gray value
        print(f"Q12: Modified pixel at ({x_coord}, {y_coord}) to 200. Saving 'modified_pixel_gray.jpg'.")

    cv2.imwrite('modified_pixel.jpg', modified_img)
    plt.figure(figsize=(6,6))
    plt.imshow(cv2.cvtColor(modified_img, cv2.COLOR_BGR2RGB)) # Convert back to RGB for display
    plt.title('Q12: Image with Modified Pixel')
    plt.axis('off')
    plt.show()

    # Q13. Read a color image and print the B, G, and R values of a selected pixel.
    if len(img.shape) == 3:
        b, g, r = img[y_coord, x_coord] # BGR order in OpenCV
        print(f"Q13: BGR values at ({x_coord}, {y_coord}): B={b}, G={g}, R={r}")
    else:
        print("Q13: Image is grayscale, cannot print B, G, R values directly.")

    # Q14. Split a color image into its B, G, and R channels and display each channel.
    if len(img.shape) == 3:
        b_channel, g_channel, r_channel = cv2.split(img)

        plt.figure(figsize=(15, 5))

        plt.subplot(1, 3, 1)
        plt.imshow(b_channel, cmap='gray')
        plt.title('Q14: Blue Channel')
        plt.axis('off')

        plt.subplot(1, 3, 2)
        plt.imshow(g_channel, cmap='gray')
        plt.title('Q14: Green Channel')
        plt.axis('off')

        plt.subplot(1, 3, 3)
        plt.imshow(r_channel, cmap='gray')
        plt.title('Q14: Red Channel')
        plt.axis('off')

        plt.show()
    else:
        print("Q14: Image is grayscale, cannot split into B, G, R channels.")

    # Q15. Merge three separate image channels into a single color image.
    if len(img.shape) == 3:
        merged_img = cv2.merge([b_channel, g_channel, r_channel])
        plt.figure(figsize=(6, 6))
        plt.imshow(cv2.cvtColor(merged_img, cv2.COLOR_BGR2RGB))
        plt.title('Q15: Merged Image (Original Color)')
        plt.axis('off')
        plt.show()
    else:
        print("Q15: Image is grayscale, merging not applicable in this context.")

    # Convert to grayscale for Q16, Q17, Q18 if not already
    if len(img.shape) == 3:
        gray_img_for_stats = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    else:
        gray_img_for_stats = img.copy()

    # Q16. Calculate and print the minimum and maximum intensity values of a grayscale image.
    min_val = np.min(gray_img_for_stats)
    max_val = np.max(gray_img_for_stats)
    print(f"Q16: Grayscale image - Min intensity: {min_val}, Max intensity: {max_val}")

    # Q17. Calculate and print the mean intensity of a grayscale image.
    mean_val = np.mean(gray_img_for_stats)
    print(f"Q17: Grayscale image - Mean intensity: {mean_val:.2f}")

    # Q18. Calculate and print the mean and standard deviation of a grayscale image using NumPy/OpenCV.
    mean_val_np = np.mean(gray_img_for_stats)
    std_dev_np = np.std(gray_img_for_stats)
    print(f"Q18: Grayscale image (NumPy) - Mean: {mean_val_np:.2f}, Standard Deviation: {std_dev_np:.2f}")

    # OpenCV also provides mean and standard deviation calculation (returns a tuple for each channel)
    mean_cv, std_dev_cv = cv2.meanStdDev(gray_img_for_stats)
    print(f"Q18: Grayscale image (OpenCV) - Mean: {mean_cv[0][0]:.2f}, Standard Deviation: {std_dev_cv[0][0]:.2f}")
```

## Q19-Q20: Image Creation (Constant Intensity & Intensity Ramp)

```python
if img is None:
    print("Error: Image not loaded. Please execute cell `7ce51561` first.")
else:
    # Q19. Create a 256 × 256 grayscale image in which every pixel has intensity value 128.
    constant_intensity_img = np.full((256, 256), 128, dtype=np.uint8)

    plt.figure(figsize=(6, 6))
    plt.imshow(constant_intensity_img, cmap='gray')
    plt.title('Q19: 256x256 Grayscale Image (Intensity 128)')
    plt.axis('off')
    plt.show()

    # Q20. Create and display a grayscale intensity ramp whose intensity gradually changes from 0 to 255.
    intensity_ramp = np.zeros((256, 256), dtype=np.uint8)
    for i in range(256):
        intensity_ramp[:, i] = i # Changes intensity horizontally from 0 to 255

    plt.figure(figsize=(6, 6))
    plt.imshow(intensity_ramp, cmap='gray')
    plt.title('Q20: Grayscale Intensity Ramp (0 to 255)')
    plt.axis('off')
    plt.show()
```

## Q21-Q25: Sampling, Quantization & Geometric Operations

```python
if img is None:
    print("Error: Image not loaded. Please execute cell `7ce51561` first.")
else:
    # Ensure we have a grayscale image for quantization tasks
    if len(img.shape) == 3:
        gray_img_for_quant = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    else:
        gray_img_for_quant = img.copy()

    # Q21. Convert an 8-bit grayscale image into a 4-bit quantized image and display the result.
    # 4-bit means 2^4 = 16 intensity levels. Each step size = 256 / 16 = 16
    quantized_4bit_img = (gray_img_for_quant // 16) * 16

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(gray_img_for_quant, cmap='gray')
    plt.title('Q21: Original Grayscale (8-bit)')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(quantized_4bit_img, cmap='gray')
    plt.title('Q21: 4-bit Quantized Grayscale')
    plt.axis('off')
    plt.show()

    # Q22. Convert an 8-bit grayscale image into a 2-bit quantized image and display the result.
    # 2-bit means 2^2 = 4 intensity levels. Each step size = 256 / 4 = 64
    quantized_2bit_img = (gray_img_for_quant // 64) * 64

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(gray_img_for_quant, cmap='gray')
    plt.title('Q22: Original Grayscale (8-bit)')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(quantized_2bit_img, cmap='gray')
    plt.title('Q22: 2-bit Quantized Grayscale')
    plt.axis('off')
    plt.show()

    # Q23. Downsample an image by a factor of 2 in both width and height and print the original and new resolutions.
    original_height, original_width = img.shape[:2]
    downsampled_width = original_width // 2
    downsampled_height = original_height // 2

    downsampled_img = cv2.resize(img, (downsampled_width, downsampled_height), interpolation=cv2.INTER_AREA)

    print(f"Q23: Original Resolution: {original_width}x{original_height}")
    print(f"Q23: Downsampled (factor 2) Resolution: {downsampled_width}x{downsampled_height}")

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title('Q23: Original Image')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(downsampled_img, cv2.COLOR_BGR2RGB))
    plt.title('Q23: Downsampled Image')
    plt.axis('off')
    plt.show()

    # Q24. Crop a rectangular Region of Interest (ROI) from an image using user-provided coordinates.
    # Example coordinates for ROI (top-left x, y, width, height)
    x_roi, y_roi, w_roi, h_roi = 100, 50, 200, 150 # Adjust as needed

    # Ensure ROI coordinates are within image bounds
    if (y_roi + h_roi <= img.shape[0]) and (x_roi + w_roi <= img.shape[1]):
        cropped_img = img[y_roi : y_roi + h_roi, x_roi : x_roi + w_roi]
        print(f"Q24: Cropped ROI from ({x_roi},{y_roi}) to ({x_roi+w_roi},{y_roi+h_roi}).")

        plt.figure(figsize=(6, 6))
        plt.imshow(cv2.cvtColor(cropped_img, cv2.COLOR_BGR2RGB))
        plt.title('Q24: Cropped Region of Interest')
        plt.axis('off')
        plt.show()
    else:
        print(f"Q24: ROI coordinates ({x_roi},{y_roi},{w_roi},{h_roi}) are out of image bounds.")

    # Q25. Rotate an image by 90 degrees, display it, and save the rotated image.
    rotated_img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

    print("Q25: Image rotated 90 degrees clockwise. Saving 'rotated_image.jpg'.")
    cv2.imwrite('rotated_image.jpg', rotated_img)

    plt.figure(figsize=(6, 6))
    plt.imshow(cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB))
    plt.title('Q25: Rotated Image (90 degrees clockwise)')
    plt.axis('off')
    plt.show()
```
